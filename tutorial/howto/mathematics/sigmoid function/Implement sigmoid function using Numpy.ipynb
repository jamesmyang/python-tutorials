# Implement sigmoid function using Numpy
https://www.geeksforgeeks.org/implement-sigmoid-function-using-numpy/

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-10, 10, 101)
z = 1 / (1 + np.exp(-x))

plt.plot(x, z)
plt.xlabel('x')
plt.ylabel('Sigmoid(x)')
```
