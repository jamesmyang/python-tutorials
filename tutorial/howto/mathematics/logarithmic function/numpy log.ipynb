```python
import numpy as np 
import matplotlib.pyplot as plt
  
in_array = [0, 1e-320, 1e-100, 0.00000000000000000001, 0.00000000001, 0.0000001, 0.0001, 0.001, 0.01, 0.1, 0.25, 0.5, 0.75, 1, np.e, 3, 5, 2**8] 
print ("Input array : ", in_array) 
  
out_array = np.log(in_array) 
print ("Output array : ", out_array) 

plt.plot(in_array, out_array)  
  
print("\nnp.log(4**4) : ", np.log(4**4)) 
print("np.log(2**8) : ", np.log(2**8)) 
```

```python
import numpy as np 
import matplotlib.pyplot as plt
  
in_array = np.linspace(0.01, 1, 100)
out_array = -np.log(in_array) 
neg_out_array = -np.log(1 - in_array) 

plt.plot(in_array, out_array, label='-log(f)')  
plt.plot(in_array, neg_out_array, label='-log(1-f)')
plt.legend()
```
