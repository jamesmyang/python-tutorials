# Machine Learning Specialization: collaborative filtering

## TensorFlow implementation

```python
import tensorflow as tf
from tensorflow import keras
```

```python
w = tf.Variable(3.0)
b = tf.Variable(0.0)
x = 1.0
y = 1.0
alpha = 0.01

iterations = 300
for iter in range(iterations):
    with tf.GradientTape() as tape:
        fwb = w * x + b
        costJ = (fwb - y)**2
    
    [djdw, djdb] = tape.gradient(costJ, [w, b])   
    
    w.assign_add(-alpha * djdw)
    b.assign_add(-alpha * djdb)
    print(costJ, w.value().numpy(), b.value().numpy())
        
w.value().numpy()
```

```python
w = tf.Variable(3.0)
b = tf.Variable(0.0)
x = 1.0
y = 1.0
alpha = 0.01

iterations = 300
for iter in range(iterations):
    with tf.GradientTape() as tape:
        #fwb = w * x + b
        costJ = (w * x + b - y)**2
    
    [djdw, djdb] = tape.gradient(costJ, [w, b])   
    
    w.assign_add(-alpha * djdw)
    b.assign_add(-alpha * djdb)
    print(costJ, w.value().numpy(), b.value().numpy())
        
w.value().numpy()
```

```python
w = tf.Variable(3.0)
b = tf.Variable(0.0)
x = 1.0
y = 1.0
alpha = 0.01
optimizer = keras.optimizers.Adam(learning_rate=alpha)

iterations = 500
for iter in range(iterations):
    with tf.GradientTape() as tape:
        #fwb = w * x + b
        costJ = (w * x + b - y)**2
    
    grads = tape.gradient(costJ, [w, b])   
    
    optimizer.apply_gradients(zip(grads, [w, b]))
    print(costJ, w.value().numpy(), b.value().numpy())
        
w.value().numpy()
```
